```python
#Fazer um jogo de adivinhação
#O usuário irá digitar um número de 0 a 10 para adivinhar o número secreto
#O usuário terá 03 tentativas para acertar o número
#O programa deverá imprimir uma mensagem se o palpite do usuário e maior ou menor que o número secreto
#O programa deverá imprimir uma mensagem de quente, morno ou frio

#Regras

#se a diferença do palpite para o número secreto for 3 - frio, 2- morno e 1 quente

import random

numero_secreto = random.randint(0, 10)

tentativas = 3

print("--- JOGO DE ADIVINHAÇÃO ---")

while tentativas > 0:
    palpite = int(input("Digite um número de 0 a 10: "))

if palpite == numero_secreto:
        print("Parabéns! Você acertou!")
        tentativas = 0

else:
        tentativas = tentativas - 1

if palpite > numero_secreto:
  print("O seu palpite é maior que o número secreto.")
else:
  print("O seu palpite é menor que o número secreto.")

if palpite > numero_secreto:
 diferenca = palpite - numero_secreto
else:
 diferenca = numero_secreto - palpite

if diferenca == 1:
  print("Dica: Está quente!")

elif diferenca == 2:
  print("Dica: Está morno!")
elif diferenca >= 3:
  print("Dica: Está frio!")


if tentativas > 0:
  print("Você ainda tem", tentativas, "tentativas.")
  print("---------------------------------")
else:
  print("---------------------------------")
  print("Fim de jogo! Suas tentativas acabaram.")
  print("O número secreto era:", numero_secreto)
```
