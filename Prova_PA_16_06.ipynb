Questão 2 - Solicite ao usuário um número inteiro positivo.

Utilizando um laço de repetição, calcule e exiba a soma de
todos os números de 1 até o número informado.

```python
numero = int(input("Digite um número: "))

soma = 0

for i in range(1, numero + 1):
    soma = soma + i

print("Resultado:", soma)
```

Questão 5 - Um caixa eletrônico possui notas de:100, 50, 20, 10, 5, 2 e 1

Solicite ao usuário o valor do saque e informe a menor
quantidade possível de notas para realizar o pagamento.

```python
valor = int(input("Digite o valor do saque: "))

nota = 100

while nota >= 1:
    quantidade = valor // nota

    if quantidade > 0:
        print(quantidade, "nota(s) de R$", nota)

    valor = valor % nota

    if nota == 100:
        nota = 50
    elif nota == 50:
        nota = 20
    elif nota == 20:
        nota = 10
    elif nota == 10:
        nota = 5
    elif nota == 5:
        nota = 2
    elif nota == 2:
        nota = 1
    else:
        break
```
