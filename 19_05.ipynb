```python
# Escrever um programa que recebe as notas dos 04 bimestre e imprimi a média com o
# resultado de aprovado, recuperação ou reprovado.

#Regras
 # Utilizar while;
 # nota igual ou maior que 7 aprovado;
 # nota igual ou maior que 5 recuperação;
 # nota menor que 5 reprovado;

nota = 0
bimestre =1
resultado = 0

while bimestre < 4:
      nota = float (input(f"digite a sua nota do {bimestre}º bimestre: "))
      bimestre +=1
      resultado = resultado + nota

if(resultado /4) >= 7:
  print(f"Sua média foi {resultado /4}, você está aprovado")
```

```python
soma = 0
contador = 1

while contador <= 4:
    nota = float(input(f"Digite a nota do {contador}º bimestre: "))
    soma += nota
    contador += 1

media = soma / 4

print(f"Média final: {media}")

if media >= 7:
    print("Você está: Aprovado")
elif media >= 5:
    print("Você está: Recuperação")
else:
    print("Você está: Reprovado")
```
